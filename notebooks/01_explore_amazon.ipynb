```python
import gzip
import json
from pathlib import Path
```

```python
review_path = Path("../data/raw/reviews_Electronics.jsonl.gz")

with gzip.open(review_path, "rt", encoding="utf-8") as f:
    line = f.readline()

review = json.loads(line)

print(review)
```

```python
print(type(review))
print(review.keys())
```
